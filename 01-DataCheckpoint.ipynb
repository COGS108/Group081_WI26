## Research Question

How did the popularity of music genres change from the pre-COVID period (2017–2019) to the post-COVID period (2022–2024), and are changes in genre popularity associated with measurable audio characteristics of tracks (tempo, energy, valence, danceability)?

## Background and Prior Work

Music has long functioned as both a cultural connector and a reflection of human emotion. The music people favor in a given era reveals not only personal taste but also wider social conditions and shared experiences. When the COVID-19 pandemic began in early 2020, it disrupted daily life globally and motivated researchers to investigate how such a large-scale crisis influenced human behavior — including how people listened to and created music — and whether those effects lasted after restrictions ended. Studies suggests that during the most intense phases of the pandemic, listeners relied on music not only for entertainment but also for emotional coping, showing increased interest in nostalgic and uplifting songs to manage stress and isolation.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

Earlier studies of pandemic-era music consumption reported notable shifts in listening patterns. For example, analyses of Last.fm streaming data showed declines in diversity and novelty of listening, alongside greater attention to familiar and popular artists, implying that uncertainty encouraged listeners to seek comfort in known music.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) Similarly, a research in the United Kingdom observed increased listening to older nostalgic tracks, especially those with positive emotional tones, indicating that both familiarity and mood influenced musical choices during lockdown periods.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

Researchers have also examined how the characteristics of music itself vary over time. Work analyzing measurable audio features — including tempo, energy, danceability, and valence — found links between these properties and listening contexts; for example, higher energy and valence tend to correspond with more active activities such as dancing and motivate the listeners to be energetic.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Another data-driven project using Spotify’s API showed that tempo, energy, and valence are among the features most strongly correlated with track popularity across genres, forming a basis for linking audio characteristics with success metrics.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4)

More recent longitudinal research has compared music trends before and after COVID-19. One large study of the top 1,000 Spotify songs from 2011 to 2023 found that pre-pandemic music showed increasing loudness and energy, whereas post-pandemic music shifted toward more acoustic and introspective qualities.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) These findings indicate that major global events can coincide with measurable changes in the sound of popular music, supporting the idea that cultural context influences musical trends.

Additional research connects these changes to broader emotional patterns in society. Sentiment analyses of song lyrics across the complete, extensive COVID-19 pandemic periods revealed themes of anxiety and introspection during COVID-19, suggesting that external stressors shape not only listening behavior but also musical creation.<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6) Overall, prior work demonstrates that music both mirrors and influences collective experience during societal disruptions, and that quantitative audio features can act as indicators of cultural mood and preference. Building on this foundation, our project investigates the relationship between genre popularity and musical characteristics during the COVID era by systematically comparing popularity patterns and audio features across defined time intervals.

```python
import pandas as pd
import numpy as np

# Load
bb = pd.read_csv("data/00-raw/billboard200.csv")

# Parse date
bb["Date"] = pd.to_datetime(bb["Date"], errors="coerce")
bb = bb.dropna(subset=["Date"]).copy()

# Keep 2017-2024
bb["Year"] = bb["Date"].dt.year
bb_17_24 = bb[bb["Year"].between(2017, 2024)].copy()

# Find latest available date in each year (year-end snapshot)
year_end_dates = (
    bb_17_24.groupby("Year")["Date"].max().sort_index()
)
print("Year-end dates:\n", year_end_dates)

# Filter to ONLY those year-end snapshots
bb_yearend = bb_17_24.merge(
    year_end_dates.rename("YearEndDate"),
    left_on="Year",
    right_index=True
)
bb_yearend = bb_yearend[bb_yearend["Date"] == bb_yearend["YearEndDate"]].copy()

print("Rows per year (should be ~200 each):")
print(bb_yearend.groupby("Year").size())

# Save what you’re using as the Billboard input for Spotify matching
bb_yearend.to_csv("data/01-interim/billboard200_yearend_2017_2024.csv", index=False)
bb_yearend.head()
```

```python
import re

def clean_song(s: str) -> str:
    if pd.isna(s): 
        return ""
    s = str(s)

    # Remove bracketed/parenthetical fluff (often "Remastered", etc.)
    s = re.sub(r"\(.*?\)", "", s)
    s = re.sub(r"\[.*?\]", "", s)

    # Remove common featuring patterns
    s = re.sub(r"\bfeat\.?.*$", "", s, flags=re.IGNORECASE)
    s = re.sub(r"\bfeaturing\b.*$", "", s, flags=re.IGNORECASE)

    # Remove trailing " - something" (remix labels)
    s = re.sub(r"\s-\s.*$", "", s)

    # Normalize whitespace
    s = re.sub(r"\s+", " ", s).strip()
    return s

def clean_artist(a: str) -> str:
    if pd.isna(a):
        return ""
    a = str(a)

    # Billboard file uses "|" sometimes; normalize to comma / "and"
    a = a.replace("|", ", ")

    # Remove featuring content in artist field too
    a = re.sub(r"\bfeat\.?.*$", "", a, flags=re.IGNORECASE)
    a = re.sub(r"\bfeaturing\b.*$", "", a, flags=re.IGNORECASE)

    a = re.sub(r"\s+", " ", a).strip()
    return a

bb_yearend["song_clean"] = bb_yearend["Song"].apply(clean_song)
bb_yearend["artist_clean"] = bb_yearend["Artist"].apply(clean_artist)

bb_yearend[["Date","Song","Artist","song_clean","artist_clean"]].head(10)
```

```python
import os
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials
from tqdm import tqdm
import time

# Recommended: set these in your environment / .env (DON'T hardcode secrets in your notebook)
# os.environ["SPOTIPY_CLIENT_ID"] = "..."
# os.environ["SPOTIPY_CLIENT_SECRET"] = "..."

sp = spotipy.Spotify(
    auth_manager=SpotifyClientCredentials(),
    requests_timeout=20,
    retries=3,
)

def spotify_search_best_track(song, artist):
    """
    Returns a dict with match info, or None if no match.
    Strategy:
      1) strict query: track:<song> artist:<artist>
      2) fallback query: <song> <artist>
    Picks best by popularity.
    """
    if not song:
        return None

    queries = []
    if artist:
        queries.append(f'track:"{song}" artist:"{artist}"')
    queries.append(f'"{song}" {artist}'.strip())

    for q in queries:
        try:
            res = sp.search(q=q, type="track", limit=5)
            items = res.get("tracks", {}).get("items", [])
            if not items:
                continue

            best = max(items, key=lambda x: x.get("popularity", 0))

            return {
                "spotify_track_id": best["id"],
                "spotify_track_name": best["name"],
                "spotify_track_popularity": best.get("popularity"),
                "spotify_explicit": best.get("explicit"),
                "spotify_duration_ms": best.get("duration_ms"),
                "spotify_preview_url": best.get("preview_url"),
                "spotify_album_id": best["album"]["id"] if best.get("album") else None,
                "spotify_album_name": best["album"]["name"] if best.get("album") else None,
                "spotify_album_release_date": best["album"].get("release_date") if best.get("album") else None,
                "spotify_primary_artist_id": best["artists"][0]["id"] if best.get("artists") else None,
                "spotify_primary_artist_name": best["artists"][0]["name"] if best.get("artists") else None,
                "spotify_search_query": q,
            }
        except Exception as e:
            # if rate-limited or transient issues, small backoff
            time.sleep(0.5)
            continue

    return None

# Run matching (with a small sleep to be gentle on rate limits)
rows = []
for i, r in tqdm(bb_yearend.iterrows(), total=len(bb_yearend), desc="Spotify search"):
    m = spotify_search_best_track(r["song_clean"], r["artist_clean"])
    out = r.to_dict()
    if m is None:
        out["spotify_matched"] = False
    else:
        out["spotify_matched"] = True
        out.update(m)
    rows.append(out)
    time.sleep(0.05)  # adjust if you hit rate limits

matched = pd.DataFrame(rows)
print("Match rate:", matched["spotify_matched"].mean())

matched.to_csv("data/01-interim/billboard200_yearend_2017_2024_with_spotify_ids.csv", index=False)
matched.head()
```
