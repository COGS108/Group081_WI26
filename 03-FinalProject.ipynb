### Abstract
This project looks into how the acoustic characteristics of popular music changed from the pre-COVID period (2017–2019) to the post-COVID period (2022). The reasoning behind this study comes from prior research suggesting that the COVID-19 pandemic influenced both music consumption and emotional expression. Because music often reflects broader societal conditions, this project examines whether these large-scale disruptions are also reflected in measurable audio features of popular songs.

To address this question, we analyzed Billboard Top-50 songs using a dataset containing waveform-derived audio features. We focused on spectral centroid (brightness), spectral rolloff (distribution of high-frequency energy), and zero crossing rate (percussiveness), which are commonly used in music information retrieval. After preprocessing the dataset, we conducted exploratory data analysis and statistical comparisons to evaluate differences in these features across the defined time periods.

Our results show that the acoustic characteristics of mainstream songs remained largely stable between the pre- and post-COVID periods. Although minor differences appear in certain feature statistics, these effects are small and not statistically significant. Additionally, classification models demonstrate weak predictive performance, indicating substantial overlap in acoustic feature space between the two periods.

In conclusion, this project suggests that while the COVID-19 pandemic may have changed the way people listen to music, it did not show many changes in the core acoustic properties of popular songs as captured by these features. These results show us the importance of considering both cultural context and methodological limitations when interpreting changes in music over time.

### Background and Prior Work
Music has long functioned as both a cultural connector and a reflection of human emotion. The music people favor in a given era reveals not only personal taste but also wider social conditions and shared experiences. When the COVID-19 pandemic began in early 2020, it disrupted daily life globally and motivated researchers to investigate how such a large-scale crisis influenced human behavior — including how people listened to and created music — and whether those effects lasted after restrictions ended. Studies suggest that during the most intense phases of the pandemic, listeners relied on music not only for entertainment but also for emotional coping, showing increased interest in nostalgic and uplifting songs to manage stress and isolation.1

Earlier studies of pandemic-era music consumption reported notable shifts in listening patterns. For example, analyses of online listening behavior using Last.fm data observed declines in diversity and novelty of listening alongside increased attention to familiar and popular artists, implying that uncertainty encouraged listeners to seek comfort in known music.2 Similarly, research also documented increased engagement with older nostalgic tracks during lockdown periods, indicating that familiarity and affective context shaped musical choices during COVID-19 disruptions.1

Researchers have also examined how the acoustic characteristics of music itself can be quantified over time using signal-processing features computed directly from audio. In music information retrieval, spectral centroid is commonly used as a descriptor of spectral shape, where higher centroid values correspond to perceptually “brighter” textures with more high-frequency content. Spectral rolloff summarizes how much energy is concentrated in lower vs higher frequencies, reflecting the distribution of spectral energy. Zero crossing rate (ZCR) measures how frequently an audio waveform changes sign and is often used as a simple indicator of noisiness or percussive activity in a signal. Together, these features provide objective, waveform-derived measurements that can be compared across songs and time periods to capture changes in timbre and high-frequency energy.3 4

Because these audio descriptors are standard in MIR, they are also frequently used in downstream tasks such as genre characterization and classification, where differences in timbral texture and spectral energy help distinguish musical styles. Classic work in automatic genre classification includes centroid, rolloff, and zero crossing measures among commonly used feature sets, motivating their use as interpretable summaries of sonic characteristics associated with different kinds of popular music.3 In applied research and reproducible pipelines, open-source audio analysis tools (e.g., librosa) implement these and related MIR features, enabling consistent extraction and comparison across large music collections.5

More recent longitudinal research shows that the timbre and loudness-related facets of popular music can exhibit measurable historical change. Large-scale analyses of contemporary Western popular music have reported patterns indicating changes in sonic palettes over time, consistent with the idea that music’s acoustic surface (timbre/texture/loudness-related attributes) evolves alongside cultural and industry contexts.6 These findings support the broader premise that major societal shifts can coincide with measurable changes in the sound of popular music.

Additional research connects pandemic-era changes to broader emotional patterns in society. Sentiment analyses of song lyrics during the COVID-19 period found increased themes of anxiety and introspection, suggesting that external stressors shape not only listening behavior but also musical creation and expression.7 Overall, prior work indicates that music both mirrors and influences collective experience during societal disruptions, and that waveform-derived audio features can serve as quantitative indicators of changes in the acoustic character of mainstream music. ### Dataset #1: Billboard Top 50 Singles Data & Audio Features (1973-2022)

This dataset contains the Billboard Top-50 singles for each year from 1973 to 2022 along with different statistics (kurtosis, max, mean, median, min, skew, std) for each in our dataset in different column, as referenced by get_features.py from the dataset. nine audio signal features extracted directly from the waveform of each song. Because the dataset links chart popularity with measurable sound characteristics, it is well suited for studying how the acoustic character of mainstream music changes across time periods such as pre- and post-COVID.

Each observation corresponds to one song in the year-end chart and includes numerical descriptors derived from digital signal processing. Unlike subjective metadata like genre labels, these variables quantify the physical properties of sound.

The variables are standardized signal-analysis measures computed from amplitude and frequency distributions. They correspond closely to human perception of music:
- Chroma STFT (Short-Time Fourier Transform): compresses audio into 12 bins representing the Western chromatic scale (C, C#, D...), regardless of octave. It calculates the energy distribution of pitch classes over time and captures harmonic progression and melody.
- Chroma CENS (Chroma Energy Normalized Statistics): refines the result of standard Chroma STFT to make data robust, smoothed, and suitable for tasks like audio matching, synchronization, and structure analysis. Usually used in comparing musical passages even when they differ in performance style, instrumentation, etc.
- RMSE (Root Mean Square Energy) – Measures average loudness/intensity of the track. Higher values indicate louder, more energetic songs. For listeners this corresponds to musical “power” or punch.
- Spectral Centroid – Represents where most frequency energy is located (low vs high frequencies). Lower values sound darker or warmer, while higher values sound brighter and sharper.
- Spectral Rolloff – Frequency below which most sound energy lies. Higher rolloff suggests more treble/high-frequency content and is often associated with energetic or modern production.
- Spectral Bandwidth – Spread of frequencies present in the audio. Larger bandwidth indicates fuller, richer instrumentation.
- Zero Crossing Rate (ZCR) – Measures how frequently the waveform changes sign. Higher values indicate more percussive or noisy sounds (e.g., electronic or heavily rhythmic tracks).
- MFCC (Mel Frequency Cepstral Coefficients) – A compact representation of timbre (“sound color”), distinguishing acoustic vs electronic textures.
- Spectral Contrast – Difference between harmonic tones and noise components, related to clarity vs distortion.

In practical terms, increases in centroid, rolloff, bandwidth, ZCR, and RMSE generally correspond to music that sounds brighter, more energetic, richer, more dynamic, and louder; while decreases correspond to softer or more mellow sound characteristics. As we focuses on the mood or emotion-related acoustic character, we focus on Spectral Centroid (brightness), Spectral Rolloff (pitch-energy), and Zero Crossing Rate (dynamic and percussion) in this project as they best reflect the brightness and dynamics of a music.

Although the dataset contains many years of music, it only includes the Top-50 most popular songs each year. Therefore, it represents mainstream commercial music rather than all music produced during that period. Niche genres and underground artists are excluded, meaning conclusions should be interpreted as describing popular music trends rather than a universal musical trend.

Another concern is that audio features are derived algorithmically from recordings rather than listener perception. For example, a song may feel emotionally positive due to lyrics or cultural links, but the dataset only captures acoustic properties. Thus the dataset measures sound characteristics, not full emotional meaning.

Additionally, production technology evolves over decades: recording quality, mastering techniques, and compression may improve every year. Some feature differences may therefore reflect improvements in recording technology rather than cultural preference. This is particularly important when comparing modern post-COVID songs to older tracks.

Finally, the dataset ignores listening frequency, streaming counts, and regional variation. A highly streamed song and a barely-ranked #50 song are treated equally, which may reduce sensitivity when analyzing popularity intensity.

#### Statistical test

```python
# Prepare numeric analysis columns
features = ['centroid_mean', 'rolloff_mean', 'zcr_mean']

for f in features:
    analysis_df[f] = pd.to_numeric(analysis_df[f], errors='coerce')

analysis_df = analysis_df.dropna(subset=features).copy()

# Inference
from scipy.stats import ttest_ind, mannwhitneyu

for f in features:
    pre = analysis_df.loc[analysis_df['Period'] == 'Pre-COVID', f].dropna()
    post = analysis_df.loc[analysis_df['Period'] == 'Post-COVID', f].dropna()

    t, p_t = ttest_ind(post, pre, equal_var=False, nan_policy='omit')
    u, p_u = mannwhitneyu(post, pre, alternative='two-sided')

    print(f'Feature: {f}')
    print('Pre mean:', pre.mean())
    print('Post mean:', post.mean())
    print('Welch t-test p-value:', p_t)
    print('Mann-Whitney p-value:', p_u)
    print()
```

### Are these changes significant? Welch’s t-test and Mann–Whitney U test
We use Welch’s t-test and Mann–Whitney U test to further spot the difference:

```python
# fun test for all variables

from scipy.stats import ttest_ind, mannwhitneyu

results = []

for col in feature_cols:
    pre = expanded_df.loc[expanded_df['Period'] == 'Pre-COVID', col].dropna()
    post = expanded_df.loc[expanded_df['Period'] == 'Post-COVID', col].dropna()

    t_stat, t_p = ttest_ind(post, pre, equal_var=False, nan_policy='omit')
    u_stat, u_p = mannwhitneyu(post, pre, alternative='two-sided')

    results.append({
        'feature': col,
        'pre_mean': pre.mean(),
        'post_mean': post.mean(),
        'difference_post_minus_pre': post.mean() - pre.mean(),
        'welch_t_stat': t_stat,
        'welch_p': t_p,
        'mannwhitney_u': u_stat,
        'mannwhitney_p': u_p
    })

results_df = pd.DataFrame(results)
```

```python
# p-value correction
from statsmodels.stats.multitest import multipletests

results_df['welch_p_fdr'] = multipletests(results_df['welch_p'], method='fdr_bh')[1]
results_df['mannwhitney_p_fdr'] = multipletests(results_df['mannwhitney_p'], method='fdr_bh')[1]

results_df = results_df.sort_values('welch_p_fdr')
```

```python
# mark significant results
results_df['welch_sig_fdr'] = results_df['welch_p_fdr'] < 0.05
results_df['mannwhitney_sig_fdr'] = results_df['mannwhitney_p_fdr'] < 0.05

# results_df[['feature', 'difference_post_minus_pre', 'welch_p', 'welch_p_fdr', 'welch_sig_fdr', 'mannwhitney_p', 'mannwhitney_p_fdr', 'mannwhitney_sig_fdr']]
```

```python
# Cohen's D
import numpy as np

def cohens_d(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    nx, ny = len(x), len(y)
    sx, sy = np.std(x, ddof=1), np.std(y, ddof=1)
    pooled = np.sqrt(((nx - 1) * sx**2 + (ny - 1) * sy**2) / (nx + ny - 1 - 1))
    return (x.mean() - y.mean()) / pooled

effect_sizes = []

for col in feature_cols:
    pre = expanded_df.loc[expanded_df['Period'] == 'Pre-COVID', col].dropna()
    post = expanded_df.loc[expanded_df['Period'] == 'Post-COVID', col].dropna()
    d = cohens_d(post, pre)
    effect_sizes.append({'feature': col, 'cohens_d': d})

effect_df = pd.DataFrame(effect_sizes)

results_df = results_df.merge(effect_df, on='feature')
results_df = results_df.sort_values('welch_p_fdr')
results_df
```

```python
# show a final result
final_results = results_df[[
    'feature',
    'pre_mean',
    'post_mean',
    'difference_post_minus_pre',
    'cohens_d',
    'welch_p',
    'welch_p_fdr',
    'mannwhitney_p',
    'mannwhitney_p_fdr'
]].sort_values('welch_p_fdr')

final_results
```

We expanded the analysis beyond mean feature values to include seven summary statistics (mean, standard deviation, skewness, kurtosis, median, minimum, and maximum) for each acoustic variable. This resulted in 21 features capturing different aspects of within-song acoustic distributions.

Initial statistical testing suggested that a small number of features, particularly spectral centroid maximum, exhibited differences between pre- and post-COVID songs. However, after applying false discovery rate (FDR) correction to account for multiple comparisons, none of the features remained statistically significant.

Additionally, effect sizes across all features were generally small (|d| < 0.3), with only a few moderate values that did not survive statistical correction.

These results indicate that any apparent differences are likely due to random variation rather than systematic changes in acoustic properties.

To verify this, we can use a logistic regression model for complementary analysis: 

### Class-balanced logistic regression model for analysis (due to the unequal sample size)

```python
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score

X = expanded_df[feature_cols]
y = expanded_df['Period']

le = LabelEncoder()
y_encoded = le.fit_transform(y)
print(le.classes_)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# model = LogisticRegression(max_iter=1000)
model = LogisticRegression(
    max_iter=2000,
    class_weight='balanced'
)
scores = cross_val_score(model, X_scaled, y_encoded, cv=5)

print("Cross-validated accuracy:", scores.mean())
print("Scores:", scores)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.model_selection import cross_val_predict

y_pred = cross_val_predict(model, X_scaled, y_encoded, cv=5)

print(confusion_matrix(y_encoded, y_pred))
print(classification_report(y_encoded, y_pred, target_names=le.classes_))
```

The Cross-validated accuracy is slightly above 0.5, indicates a measurable acoustic shift, but being less than ideal.

As Post-COVID has an index of 0 and Pre-COVID has an index of 1, a positive coefficient from the logistic regression model indicates a higher value in general for pre-COVID era, and a negative coefficient for the opposite (higher value for post-COVID era).

```python
# find which acoustic summaries matter most in the model
model.fit(X_scaled, y_encoded)

coef_df = pd.DataFrame({
    'feature': feature_cols,
    'coefficient': model.coef_[0],
    'abs_coefficient': np.abs(model.coef_[0])
}).sort_values('abs_coefficient', ascending=False)

coef_df.head(10)
```

The top model coefficients suggests a strong clue that the differences are being driven less by average level and more by internal shape, peaks, and variability. However, the Cross-validated accuracy indicates that these differences also seems insignificant.
